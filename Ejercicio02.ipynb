````python
n = int(input('Ingrese el valor de n: '))
sum = 0
print('Numeros pares hasta n: ', end='')
for i in range(1, n+1):
    sum += i #Calcula la suma de los primeros n números naturales.
    if i!= 0 and i%2 == 0: #Imprime solo los números pares usando if.
      print(i, end=' ' )
print(f'\nLa suma de los naturales hasta {n} es: {sum}')
````

````python
#Usando el ejercicio anterior de las temperaturas, encuentra las temperaturas mayores al promedio
# Ejercicio 1.2
# Lista inicial de temperaturas en °C
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
#Muestra la lista ordenada de menor a mayor.
temperaturas.sort()
print("Temperaturas:", temperaturas)
#Calcula el promedio de todas las temperaturas.
promedio = sum(temperaturas)/len(temperaturas)
print("Promedio:", promedio)
#Calcula mayores al promedio
print("Temperaturas mayores al promedio: ", end='')
for temperatura in temperaturas:
  if temperatura > promedio:
    print(temperatura, end=' ')
````

````python
#Crea un programa que solicite números hasta que el usuario ingrese 0.
n=1
print('El programa se detendrá cuando se ingrese "0"')
while n != '0':
  n = input('Ingrese un número: ')
print('Programa finalizado.')
````

**Estado del ciclo:**

En el código, el estado del ciclo está representado principalmente por la variable n y la condición del while:

`while n != '0':`

1. Variable n:

Guarda el último valor ingresado por el usuario.
Este valor determina si el ciclo sigue ejecutándose o se detiene.

2. Condición n != '0':

Es la regla que define si el ciclo está activo o finalizado.
Mientras n sea diferente de '0', el ciclo sigue en estado activo.
Cuando n es '0', el estado pasa a terminado y el while deja de ejecutarse.

3. Interacción del usuario:

Cada entrada modifica el estado del ciclo, ya que cambia el valor de n.









### **2.4**

````python
suma_pares = 0
suma_impares = 0 #Agrega otra variable para acumular los impares.
for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i
print("Suma de los pares:", suma_pares)
print('Suma de los impares:', suma_impares)
#Representa el estado final como un diccionario (suma_pares y suma_impares)
estado_final = [
    {'Suma de pares': suma_pares},
    {'Suma de impares': suma_impares}
]
print(estado_final)
````

**Cambio de estado:**

En el código, el estado del ciclo cambia en cada iteración del `for` porque las variables `suma_pares` y `suma_impares` se van actualizando según el valor de `i`. Al final, el estado queda fijo y se imprime como un diccionario.

Inicio del ciclo:
```
suma_pares = 0
suma_impares = 0
```
Estado inicial: `{'Suma de pares': 0, 'Suma de impares': 0}`

Iteraciones del for:

El ciclo recorre `i` desde 1 hasta 10.
En cada iteración:

Si `i` es par `(i % 2 == 0)`, se suma a `suma_pares`.
Si `i` es impar, se suma a `suma_impares`.

Esto modifica el estado del ciclo porque las variables acumuladoras cambian.

Estado final:
`{'Suma de pares': 30, 'Suma de impares': 25}`

Esto se representa en el código como:
```
estado_final = [
    {'Suma de pares': suma_pares},
    {'Suma de impares': suma_impares}
]
```










### **2.5**

````python
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))
    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

# Mostrar total
total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]

print("Total del inventario:", total)
````
