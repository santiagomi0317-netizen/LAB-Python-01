```python
contador_global = 0

def incrementar():
    contador_global = contador_global + 1

incrementar()
print(contador_global)
```

En Python, al asignar un valor a una variable dentro de una función `contador_global = ...`, el intérprete asume que es una variable local. Como se intenta sumar 1 a esa variable antes de que tenga un valor asignado localmente, el programa falla.

```python
contador_global = 0

def incrementar():
    global contador_global
    contador_global = contador_global + 1

incrementar()
print(contador_global) # Imprime 1
```

```python
contador_global = 0

def incrementar(contador):
    return contador + 1

contador_global = incrementar(contador_global)
print(contador_global) # Imprime 1
```

La Solución 2 es mucho más fácil de probar porque es una función "pura". Su resultado depende únicamente del parámetro que se le pasa, no de variables ocultas en el entorno global. Esto evita efectos secundarios indeseados y facilita la depuración.

```python
def agregar_nota(notas, nueva):
    notas.append(nueva)

def reiniciar(notas):
    notas = []

mis_notas = [3.5, 4.0]
agregar_nota(mis_notas, 4.8)
reiniciar(mis_notas)
print(mis_notas)
```

Predicción: Al ejecutar el código, se imprimirá [3.5, 4.0, 4.8]. La función reiniciar no afectará la lista original.

Explicación: Modificar un objeto mutable `append()` cambia el espacio en memoria original donde vive la lista. En cambio, reasignar `notas = []` simplemente crea una nueva lista vacía en otro espacio de memoria y hace que la variable local notas apunte a ella, dejando la lista original intacta.

```python
def agregar_nota(notas, nueva):
    print(f"  [agregar_nota] ID notas antes: {id(notas)}")
    notas.append(nueva)

def reiniciar(notas):
    print(f"  [reiniciar] ID notas antes: {id(notas)}")
    notas.clear() # Vacía la lista original sin reasignar
    print(f"  [reiniciar] ID notas después: {id(notas)}")

mis_notas = [3.5, 4.0]
print(f"[Global] ID mis_notas: {id(mis_notas)}")

agregar_nota(mis_notas, 4.8)
reiniciar(mis_notas)

print(mis_notas) # Ahora sí imprime []
```

```python
#Implementa las funciones consultar(cuenta), depositar(cuenta, monto) y retirar(cuenta, monto)
def consultar(cuenta):
    print(f"\n--- Saldo actual: ${cuenta['saldo']} ---")

def depositar(cuenta, monto):
    if monto > 0:
        cuenta["saldo"] = cuenta["saldo"] + monto
        cuenta["movimientos"].append(("Depósito", monto, cuenta["saldo"])) #Cada operación exitosa debe registrar un movimiento en la lista movimientos (tipo, monto, saldo resultante).
        print(f"\nDepósito de ${monto} exitoso.")
    else:
        print("\nError: El monto a depositar debe ser mayor a 0.")

def retirar(cuenta, monto):
    if monto <= 0:
        print("\nError: El monto a retirar debe ser mayor a 0.") #
    elif cuenta["saldo"] >= monto:
        cuenta["saldo"] = cuenta["saldo"] - monto
        cuenta["movimientos"].append(("Retiro", monto, cuenta["saldo"]))
        print(f"\nRetiro de ${monto} exitoso.")
    else:
        print("\nError: Fondos insuficientes.") #retirar no debe permitir dejar el saldo en negativo; en ese caso imprime un mensaje y no modifica el estado.

cuenta = {"titular": "Ana", "saldo": 200_000, "movimientos": []}

while True: #Construye un menú con while que permita elegir la operación hasta que el usuario seleccione "salir".
    print("\n1. Consultar saldo\n2. Depositar\n3. Retirar\n4. Salir")
    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        consultar(cuenta)
    elif opcion == "2":
        monto = float(input("Ingrese monto a depositar: "))
        depositar(cuenta, monto)
    elif opcion == "3":
        monto = float(input("Ingrese monto a retirar: "))
        retirar(cuenta, monto)
    elif opcion == "4":
        print("\n=== EXTRACTO DE MOVIMIENTOS ===") #Al salir, muestra el extracto completo de movimientos en forma de tabla.
        print("TIPO\t\tMONTO\t\tSALDO")
        print("-" * 45)
        for mov in cuenta["movimientos"]:
            print(f"{mov[0]}\t${mov[1]}\t\t${mov[2]}")
        break
    else:
        print("\nOpción inválida.")
```
