```python
#Ejercicio 1
# Estado inicial
saldo = float(input("ingrese su saldo:"))   # Saldo de la cuenta
deposito = float(input("ingrese su deposito:"))   # Monto a depositar
retiro = float(input("ingrese su retiro:"))     # Monto a retirar

# Actualización del estado
saldo = saldo + deposito   # Se deposita dinero
saldo = saldo - retiro     # Se retira dinero

# Salida del estado final
print("Saldo final de la cuenta:", saldo)

interes = 0.02
saldo_final = saldo + saldo*interes

print(f"Saldo final con interés: ${saldo_final:.2f}")
```

*   ¿Qué representa el estado inicial y el estado final del programa?
El estado inicial representa las variables ya instancidas  con los cuales vamos a a trabajar, (datos ya dados). Mientras que el estado final resulta de realziar ciertas operaciones a nuestras variables del inicio para llegar a un resultado esperado
*   Explica cómo cada instrucción afecta la memoria.

1.   Asignacion: Este crea el objeto del tipo del valor asignado y apunta la variable apunte a este espacio de memoria aumentado del contador de referencias.
2.   Tipos inmutables(int, str, tuple): al realizar cambios en una variable este crea un nuevo objeto con el nuevo valor,asi mismo su apuntador de memoria se actualiza al nuevo objeto
3.   Tipos Mutables(list, dict, set): Este no cambio su valor de memoria, solo aumenta su tamaño en bytes
4. def funcion(): python crea un objeto de tipo funcion y su memoria permanece viva mientras este corriendo su contenido.
5. class a: python crea un objeto de tipo clase, guarda todos sus metodos en memoria y cuando se instancia, ese objeto tiene un espacio de memoria dedicado a sus atributos
6. if / else: NO genera estrucuturas adicionales de momeria
7. for i in range(1000): solo guarda en memoria el generador i

```python
#Ejercicio 3
# Lista de productos en una tienda
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 7},
    {"nombre": "Café", "precio": 8000, "stock": 9},]

# Muestra el nombre del primer producto.
print(inventario[0]["nombre"])

#Agrega un nuevo producto al inventario.
inventario.append({"nombre": "Huevos", "precio": 567.543, "stock": 180})
inventario.append({"nombre": "nada", "precio": 0 , "stock": 0})
print(inventario)

#Disminuye el stock de “Leche” en 2 unidades (venta).
inventario[1]["stock"] -= 2
print(inventario)

#Calcula el valor total del inventario (precio * stock por producto).
total = 0
for rep in range(len(inventario)):
  total += inventario[rep]["precio"] * inventario[rep]["stock"]
print(total)

#Muestra los productos cuyo stock sea menor a 10.
print("Los productos cuyo stock es menor a 10 son", end=" ")
for rep in range(len(inventario)):
  if inventario[rep]["stock"]<10:
    print(inventario[rep]["nombre"],end=" ")

#Aumenta el precio de todos los productos en un 10%.
for rep in range(len(inventario)):
  inventario[rep]["precio"] += inventario[rep]["precio"] * 0.1
print("")
print(inventario)

#Redondea los precios con round(p["precio"], 2) para evitar decimales largos.
for rep in range(len(inventario)):
  inventario[rep]["precio"] = round(inventario[rep]["precio"], 2)
print(inventario)

#Busca un producto por nombre (ingresado por el usuario) y muestra el producto encontrado, ten en cuenta que no puede estar.
band= False
nombre= input("Ingrese el nombre del producto")
for rep in range(len(inventario)):
  if nombre == inventario[rep]["nombre"]:
    print(inventario[rep])
    band = True
    break
if not band:
  print("Elemento no encontrado")

#Elimina un producto del inventario si su stock es 0.
for rep in range(len(inventario)):
  if 0 == inventario[rep]["stock"]:
    inventario.pop(rep)
print(inventario)

#Ordena la lista por precio ascendente.
def obtener_precio(producto):
    return producto["precio"]
inventario.sort(key=obtener_precio)
print(inventario)

```

```python
#Ejercicio 4
from math import sqrt

# Tupla: coordenadas (x, y)
punto = (3, 7)
print("Punto:", punto)

#Crea una tupla rectangulo = (ancho, alto).
rectangulo = (5,5)
print("Rectangulo:", rectangulo)

#Calcula su área y perímetro.
area = rectangulo[0] * rectangulo[0]
print(area)

#Calcula la distancia total recorrida dadas las posiciones anteriores
def distancia(punto1,punto2):
  return sqrt(((punto2[0]-punto1[0])**2)+ ((punto2[1]-punto1[1])**2))

posiciones = [(0, 0), (1, 2), (3, 4)]
total = 0
for rep in range(len(posiciones)-1):
  total += distancia(posiciones[rep],posiciones[rep+1])
print(total)
```

```python
#Ejercicio 5
biblioteca = {
    "libros": [{"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True},],
    "usuarios": ["Ana", "Luis", "María"]
}

#Muestra cuántos libros están prestados y cuántos disponibles.
prestados = 0
disponibles = 0
for libro in biblioteca["libros"]:
    if libro["prestado"]:
        prestados += 1
    else:
      disponibles += 1
print(f"Libros prestados: {prestados}")
print(f"Libros disponibles: {disponibles}")

#Agrega un nuevo libro.

biblioteca["libros"].append({
    "titulo": "Cien años de soledad",
    "autor": "Gabriel García Márquez",
    "prestado": False})

print(biblioteca["libros"])

#Marca como prestado el libro “1984”.
for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True
        break
print(biblioteca["libros"])

#Muestra el estado actualizado del sistema (usa print(biblioteca)).
print(biblioteca)

```
