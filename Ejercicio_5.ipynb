```python
#Ejercicio 1
import random

MAX_INTENTOS = 7
ganadas = 0
perdidas = 0

print("Juego de adivinanza")

while True:
    secreto = random.randint(1, 100)
    historial = []
    acerto = False

    print(f"\nHe pensado un número entre 1 y 100. Tienes {MAX_INTENTOS} intentos.")

    while len(historial) < MAX_INTENTOS:
        restantes = MAX_INTENTOS - len(historial)
        entrada = int(input(f"Intento {len(historial) + 1} ({restantes} restantes): "))

        # Validación: no consume intento si la entrada es inválida
        if not (1 <= entrada <= 100):
            print("Ingresa un número entero entre 1 y 100.")
            continue


        historial.append(entrada)

        if entrada == secreto:
            acerto = True
            break
        elif secreto > entrada:
            print("El número es mayor")
        else:
            print("El número es menor")


    if acerto:
        ganadas += 1
        print(f"\n Correcto! Lo adivinaste en {len(historial)} intentos.")
    else:
        perdidas += 1
        print(f"\n Perdiste. El número era {secreto}.")


    print("\nHistorial de intentos")
    for i in range(len(historial)):
      print(f"  {i + 1}. {historial[i]}")

    # Marcador
    print(f"\n Marcador  Ganadas: {ganadas} | Perdidas: {perdidas}")

    # ¿Jugar otra vez?
    while True:
        respuesta = input("\n¿Quieres jugar otra vez? (s/n): ").lower()
        if respuesta in ("s", "n"):
            break
        print("Responde 's' o 'n'.")

    if respuesta == "n":
        print("\n¡Gracias por jugar!")
        break
```

¿Por qué 7 intentos son suficientes si se juega de forma óptima? Relaciónalo con la búsqueda binaria del Ejercicio 5.
Al partir cada rato la lista de los 100 numeros a la mitad(busqueda binaria), esta tiene una complejiad de O(logn), si hacemos la particion de tal manera, 100 → 50 → 25 → 13 → 7 → 4 → 2 → 1 vemos que en el septimo paso ya dedujimos un unico valor que es el que tenemos que adivinar

```python
#Ejercicio 1
FILAS = 10
COLUMNAS = 10
VIVA = 1
MUERTA = 0

def crear_tablero():
  tablero = []
  for i in range(FILAS):
    fila = []
    for j in range(COLUMNAS):
        fila.append(MUERTA)
    tablero.append(fila)
  return tablero

def contar_vecinos(tablero, fila, col):
  vecinos = 0
  for df in range(-1, 2):
    for dc in range(-1, 2):
      if df == 0 and dc == 0:
         continue

      f = fila + df
      c = col + dc

      if 0 <= f < FILAS and 0 <= c < COLUMNAS:
        vecinos += tablero[f][c]
  return vecinos

def siguiente_generacion(tablero):
  nuevo = crear_tablero()
  for i in range(FILAS):
    for j in range(COLUMNAS):
        vecinos = contar_vecinos(tablero, i, j)
        if tablero[i][j] == VIVA:
            if vecinos == 2 or vecinos == 3:
                nuevo[i][j] = VIVA
            else:
                nuevo[i][j] = MUERTA
        else:
            if vecinos == 3:
                nuevo[i][j] = VIVA
            else:
                nuevo[i][j] = MUERTA

  return nuevo

def mostrar(tablero, generacion):
  print(f"\nGeneración {generacion}")
  for fila in tablero:
    print(fila)

tablero = crear_tablero()
tablero[0][4] = VIVA
tablero[1][4] = VIVA
tablero[2][4] = VIVA
tablero[2][3] = VIVA
tablero[2][2] = VIVA

for generacion in range(1, 6):
  mostrar(tablero, generacion)
  tablero = siguiente_generacion(tablero)
```

Explica por qué es necesario usar un tablero nuevo y no modificar el actual mientras se recorre.
Al tener de entrada el estado anterior, necesitamos que todos su valores correspondan a este estado, Si solo manejamos un tablero miestras lo recorremos podemos modificar casillas las cuales no corresponda al tablero anterior y estariamos llegando a estados incorrectos.
