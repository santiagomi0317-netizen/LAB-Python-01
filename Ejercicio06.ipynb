```python
#Toma el programa de inventario del Ejercicio 2.5 y protege la lectura de precio y cantidad con try / except ValueError.
productos = []

while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre.lower() == "fin":
        break

    while True: #Si el usuario ingresa un valor inválido, vuelve a pedírselo con un while hasta que sea correcto.
        try:
            precio = float(input("Precio: "))
            cantidad = int(input("Cantidad: "))
            if precio < 0 or cantidad < 0:
                print("Error: El precio y la cantidad no pueden ser negativos.")
                continue
        except ValueError:
            print("Error: Ingrese valores numéricos válidos.")
        else: #Usa else y finally en algún bloque try y explica cuándo se ejecuta cada uno.
            #Se ejecuta SOLO si el bloque try funcionó sin levantar excepciones
            productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})
            print("Producto registrado correctamente.")
            break
        finally:
            #Se ejecuta SIEMPRE, haya ocurrido un error o no
            print("-> Intento de registro procesado.\n")

print("\nInventario final:", productos)
```

```python
#Guarda el inventario en un archivo inventario.csv usando open(..., "w") y escribiendo línea por línea.
def guardar_csv(inventario):
    archivo = open("inventario.csv", "w") #
    for p in inventario:
        archivo.write(f"{p['nombre']},{p['precio']},{p['cantidad']}\n")
    archivo.close()
#Crea una función que lea el archivo con open(..., "r"), recorra sus líneas y reconstruya la lista de diccionarios.
def cargar_csv():
    inventario = []
    try:
        archivo = open("inventario.csv", "r")
        for linea in archivo:
            # Reconstrucción manual limpiando saltos de línea y separando por comas
            datos = linea.strip().split(",")
            inventario.append({
                "nombre": datos[0],
                "precio": float(datos[1]),
                "cantidad": int(datos[2])
            })
        archivo.close()
    except FileNotFoundError: #Maneja el caso en que el archivo no exista (FileNotFoundError) iniciando con un inventario vacío.
        print("Archivo CSV no encontrado. Iniciando inventario vacío.")
    return inventario
```

```python
#Repite el ejercicio usando el módulo json (json.dump / json.load). ¿Qué ventajas tiene frente al CSV?
import json

def guardar_json(inventario):
    archivo = open("inventario.json", "w")
    json.dump(inventario, archivo)
    archivo.close()

def cargar_json():
    inventario = []
    try:
        archivo = open("inventario.json", "r")
        inventario = json.load(archivo)
        archivo.close()
    except FileNotFoundError:
        print("Archivo JSON no encontrado. Iniciando inventario vacío.")
    return inventario
```

Ventajas del JSON frente al CSV: El formato JSON preserva de forma nativa los tipos de datos. Con CSV, toda la información se lee como texto, obligando al programador a realizar conversiones manuales y a tener cuidado con los separadores si el nombre de un producto ya contiene una coma. Con `json.load()`, la estructura de datos se reconstruye en Python exactamente igual a como se guardó.
